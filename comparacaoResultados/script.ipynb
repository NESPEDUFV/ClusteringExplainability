```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python

def plot_metrics_by_cluster(csv_file):

    df = pd.read_csv(csv_file)

    clusters = df['Cluster'].unique()
    metrics = ['Coverage', 'Separation Error', 'Conciseness']

    plt.figure(figsize=(15, 10))

    for i, cluster in enumerate(clusters, start=1):
        plt.subplot(2, 2, i)
        
        subset = df[df['Cluster'] == cluster]
        
        bar_width = 0.2
        x = range(len(subset['Algoritmo']))
        
        plt.bar(x, subset['Coverage'], width=bar_width, label='Coverage', alpha=0.6)
        plt.bar([p + bar_width for p in x], subset['Separation Error'], width=bar_width, label='Separation Error', alpha=0.6)
        plt.bar([p + bar_width * 2 for p in x], subset['Conciseness'], width=bar_width, label='Conciseness', alpha=0.6)

        plt.title(f'Cluster {cluster}')
        plt.xlabel('Algoritmos')
        plt.ylabel('Valores das Métricas')
        plt.xticks([p + bar_width for p in x], subset['Algoritmo'], rotation=90)
        plt.grid()

    plt.tight_layout()
    plt.show()
```

```python
plot_metrics_by_cluster('./csv/metrics_output_iris.csv')
```

```python
#wine
plot_metrics_by_cluster('./csv/metrics_output_wine.csv')
```
